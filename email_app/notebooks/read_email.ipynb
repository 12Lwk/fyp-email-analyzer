```python
import marimo as mo
from gmail_api import init_gmail_service, get_email_messages, get_email_message_details

client_file = 'C:/Users/User/OneDrive - Asia Pacific University/APU Final Year Project/FYP Email Project Documents/Email Dataset Python/GMAIL_API/credentials.json'
service = init_gmail_service(client_file)
```

```python
messages = get_email_messages(service, folder_name='INBOX')
```

```python
messages
```

```python
import csv

csv_filename = "personal_gmail_data.csv"

with open(csv_filename, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Subject", "From", "Recipients", "Body", "Snippet", "Has Attachments", "Date", "Star", "Label"])

    #Loop through messages and write data to CSV
    for msg in messages:
        details = get_email_message_details(service, msg['id'])

        if details:
            writer.writerow([
                details['subject'],
                details['sender'],
                details['recipients'],
                details['body'][:100],  #Store first 100 characters of the body
                details['snippet'],
                details['has_attachments'],  
                details['date'],
                details['star'],
                details['label']
            ])

print(f"Email data has been saved to {csv_filename}")
```
